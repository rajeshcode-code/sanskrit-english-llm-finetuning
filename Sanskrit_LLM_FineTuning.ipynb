```python
#Load the base Qwen model
import torch
from transformers import AutoModelForCausalLM

MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="auto"
)

print("Model loaded successfully")
print("Model device:", model.device)
```

```python

from transformers import AutoTokenizer

MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("Model:", MODEL_NAME)
print("Tokenizer loaded successfully")
print("Vocabulary size:", tokenizer.vocab_size)
```

```python
def generate_response(prompt, max_new_tokens=150):
    messages = [
        {"role": "user", "content": prompt}
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False
        )

    generated_tokens = outputs[0][inputs["input_ids"].shape[1]:]

    return tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )


prompt = "Translate this Sanskrit sentence into English: आयुर्वेदः नाम आयुः वेदयति इति आयुर्वेदः।"

response = generate_response(prompt)

print("PROMPT:")
print(prompt)

print("\nMODEL RESPONSE:")
print(response)
```

```python
import sacrebleu

test_sentences = [
    "त्रयो दोषाः वातः पित्तं कफः इति",
    "वातः गतिः कारणम्",
    "व्यायामः शारीरस्य बलं वर्धनं करोति",
    "रोगः नाम दोष वैषम्य कारणम्"
]

references = [
    "There are three doshas: Vata, Pitta, and Kapha.",
    "Vata is responsible for movement.",
    "Exercise increases the strength of the body.",
    "Disease is caused by an imbalance of the doshas."
]

base_outputs = []
fine_outputs = []

for sentence in test_sentences:
    prompt = f"Translate this Sanskrit sentence into English: {sentence}"

    with model.disable_adapter():
        base_outputs.append(generate_response(prompt))

    fine_outputs.append(generate_response(prompt))

base_bleu = sacrebleu.corpus_bleu(base_outputs, [references]).score
fine_bleu = sacrebleu.corpus_bleu(fine_outputs, [references]).score

print("BASE MODEL BLEU:", round(base_bleu, 2))
print("FINE-TUNED MODEL BLEU:", round(fine_bleu, 2))
print("BLEU improvement:", round(fine_bleu - base_bleu, 2))
```

```python
import pandas as pd

evaluation_df = pd.DataFrame({
    "sanskrit": test_sentences,
    "reference": references,
    "base_model": base_outputs,
    "fine_tuned_model": fine_outputs
})

evaluation_df.to_csv(
    "sanskrit_translation_evaluation.csv",
    index=False
)

print("Evaluation results saved successfully")
print(evaluation_df)
```

```python
# Error analysis

for i in range(len(evaluation_df)):
    reference = evaluation_df.loc[i, "reference"]
    base = evaluation_df.loc[i, "base_model"]
    fine = evaluation_df.loc[i, "fine_tuned_model"]

    print("=" * 70)
    print("Example", i + 1)
    print("Reference:", reference)

    print("\nBase model output:")
    print(base)

    print("\nFine-tuned output:")
    print(fine)

    print("\nObservation:")
    if len(base) > len(reference) * 2:
        print("- Base model produces substantially more text than the reference.")
    if len(fine) > len(reference) * 2:
        print("- Fine-tuned model produces substantially more text than the reference.")
    if "translation" in base.lower() or "this refers" in base.lower():
        print("- Base model adds explanatory/meta text beyond the translation.")
    if "translation" in fine.lower() or "this refers" in fine.lower():
        print("- Fine-tuned model adds explanatory/meta text beyond the translation.")
```
